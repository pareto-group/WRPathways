```python
import glob
from pathlib import Path
import xarray as xr
```

```python
def cut_img(ds):
    latSlice = slice(80, 20) #20N, 80N
    lonSlice = slice(180, 330) #180W, 30W
    return ds.sel(latitude=latSlice, longitude=lonSlice).sel(
        level=500.).resample(time='D').mean(skipna=True)
```

```python
out_dir = '/glade/derecho/scratch/molina/e5_uwind/'
```

```python
list_u = sorted(
    glob.glob(
        '/glade/campaign/collections/gdex/data/d633000/e5.oper.an.pl/*/e5.oper.an.pl.128_131_u.ll025uv.*.nc'
    )
)
```

```python
for fi in list_u[3493:3653]:
    out_path = out_dir+Path(fi).name
    ds = xr.open_mfdataset(fi, preprocess=cut_img, parallel=False)['U']
    ds.to_netcdf(out_path)
    print(out_path)
```
